Combine all saved batches

```python
container_client = blob_service.get_container_client("bronze")
batch_blobs = sorted(container_client.list_blobs(name_starts_with="live/batches/"), key=lambda b: b.name)

print(f"Found {len(batch_blobs)} batch files")

all_batches = []
for blob in batch_blobs:
    blob_client = blob_service.get_blob_client(container="bronze", blob=blob.name)
    stream = blob_client.download_blob().readall()
    data = json.loads(stream)
    batch_df = pd.DataFrame(data.get("value", data)) if isinstance(data, dict) else pd.DataFrame(data)
    if "value" in batch_df.columns and len(batch_df) == 1:
        batch_df = pd.DataFrame(batch_df["value"].iloc[0])
    all_batches.append(batch_df)
    print(f"  {blob.name}: {len(batch_df)} rows")

full_bronze = pd.concat(all_batches, ignore_index=True).drop_duplicates()
print(f"\nCombined total: {len(full_bronze)}")
print(f"Date range: {pd.to_datetime(full_bronze['postingDate']).min()} to {pd.to_datetime(full_bronze['postingDate']).max()}")
```
