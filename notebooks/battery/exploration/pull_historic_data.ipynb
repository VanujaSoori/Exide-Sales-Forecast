```python
from azure.storage.blob import BlobServiceClient
import json
import pandas as pd
import io

conn_str = f"DefaultEndpointsProtocol=https;AccountName={storage_account_name};AccountKey={storage_account_key};EndpointSuffix=core.windows.net"
blob_service = BlobServiceClient.from_connection_string(conn_str)
```

```python
import requests

headers = {"Authorization": f"Bearer {access_token}", "Accept": "application/json"}

all_records = []
url = erp_api_url
page = 1
while url:
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    data = response.json()
    all_records.extend(data.get("value", []))
    print(f"Page {page}: {len(data.get('value', []))} records (total: {len(all_records)})")
    url = data.get("@odata.nextLink")
    page += 1

full_bronze = pd.DataFrame(all_records)
print(full_bronze.shape)
```

```python
blob_client = blob_service.get_blob_client(container="bronze", blob="live/battery_full_history.json")
stream = blob_client.download_blob().readall()
data = json.loads(stream)

full_bronze = pd.DataFrame(data.get("value", data)) if isinstance(data, dict) else pd.DataFrame(data)
print(f"Full history: {full_bronze.shape}")
print(f"Date range: {pd.to_datetime(full_bronze['postingDate']).min()} to {pd.to_datetime(full_bronze['postingDate']).max()}")
```
